```python
from sklearn import datasets
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
```

```python
x = np.load(r'preprocessed_output\Sub1_Ses1_ALL_X.npy').reshape(-1, 32 * 308)
y = np.load(r'preprocessed_output\Sub1_Ses1_ALL_y.npy')

print(x.shape)
print(y.shape)
```

```python
print(x[0])
print(y[0])
```

```python
print('x shape:', x.shape)        # (trials, channels, timepoints)
print('y shape:', y.shape)        # (trials,)
print('Unique image IDs:', len(np.unique(y)))  # how many distinct images
print('Sample y values:', y[:5])  # first 5 image labels
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, 
                                                    test_size=0.2,
                                                    random_state=432)
```

```python
pca = PCA(n_components=4000)
x_train_pca = pca.fit_transform(x_train)
x_test_pca = pca.transform(x_test)
```

```python
model = LinearDiscriminantAnalysis()
model.fit(x_train_pca, y_train)
y_pred = model.predict(x_test_pca)
accuracy = accuracy_score(y_test, y_pred)
```
